```python
df_data = pd.read_csv("2026VAERSDATA.csv", encoding="latin1", low_memory=False)
df_vax = pd.read_csv("2026VAERSVAX.csv", encoding="latin1", low_memory=False)
df_symptoms = pd.read_csv("2026VAERSSYMPTOMS.csv", encoding="latin1", low_memory=False)

print("Initial shapes:")
print("  2026VAERSDATA.csv   :", df_data.shape)
print("  2026VAERSVAX.csv    :", df_vax.shape)
print("  2026VAERSSYMPTOMS.csv:", df_symptoms.shape)
```

<div style="background-color: #f8fafc; border: 1px solid #cbd5e1; border-radius: 6px; padding: 14px 18px; font-family: 'Segoe UI', Arial, sans-serif; color: #1e293b; margin-top: 6px;">
  <h4 style="color: #14532d; margin: 0 0 8px 0; font-size: 15px;">Explanation — Step 2: Loading the 2026 VAERS CSV Files</h4>
  <ul style="margin: 0; padding-left: 20px; line-height: 1.65; font-size: 14px;">
    <li><strong>What this step does:</strong> Reads the patient, vaccine, and symptom CSV files into separate DataFrames and prints their shapes.</li>
    <li><strong>Why these choices were made:</strong> <code>encoding="latin1"</code> prevents character-decoding errors in free-text fields, and <code>low_memory=False</code> avoids mixed-type chunk warnings across large files.</li>
    <li><strong>Interpretation:</strong> The printed shapes confirm the exact row and column counts of each raw source table before merging.</li>
  </ul>
</div>

<div style="background-color: #fefce8; border-left: 5px solid #ca8a04; border-radius: 6px; padding: 12px 16px; font-family: 'Segoe UI', Arial, sans-serif; color: #1e293b;">
  <h3 style="color: #713f12; margin: 0; font-size: 18px;">
    Step 3: Merging Patient, Vaccine &amp; Symptom Tables on VAERS_ID
  </h3>
  <p style="margin: 4px 0 0 0; font-size: 13px; color: #475569;">
    Combine the three tables step-by-step using VAERS_ID and inspect the unified DataFrame dimensions and columns.
  </p>
</div>

```python
merged_data_vax = pd.merge(df_data, df_vax, on="VAERS_ID", how="inner")
df_merged = pd.merge(merged_data_vax, df_symptoms, on="VAERS_ID", how="inner")

print("\nMerged dataset shape:", df_merged.shape)
print("Total columns in merged dataset:", len(df_merged.columns))
print("\nColumn names:")
print(list(df_merged.columns))
```

```python
# 1. Fill symptom columns with 'None'
for i in range(1, 6):
    sym_col = f"SYMPTOM{i}"
    if sym_col in df_clean.columns:
        df_clean[sym_col] = df_clean[sym_col].fillna("None")
```

```python
# 2. Check which columns still have missing values before filling them
still_null_cols = df_clean.columns[df_clean.isnull().sum() > 0].tolist()
print("Remaining columns with nulls before Step 5F fill:", still_null_cols)
```

```python
# Convert RECVDATE to proper datetime format
if "RECVDATE" in df_clean.columns:
    df_clean["RECVDATE"] = pd.to_datetime(df_clean["RECVDATE"], errors="coerce")
    print("RECVDATE dtype:", df_clean["RECVDATE"].dtype)
```

    We format <code>RECVDATE</code> as a proper date column, ensure <code>AGE_YRS</code>, <code>NUMDAYS</code>, and <code>HOSPDAYS</code> are numeric (filling any conversion blanks), and set any negative numbers to <code>0</code> so calculations and charts stay accurate.

<div style="background-color: #fef2f2; border-left: 5px solid #dc2626; border-radius: 6px; padding: 10px 16px; font-family: 'Segoe UI',Arial, sans-serif; color: #1e293b;">
  <h3 style="color: #7f1d1d; margin: 0 0 6px 0; font-size: 18px;">Step 8: Handling Extreme Outliers Using IQR (Capping)</h3>
</div>

```python
# Check maximum NUMDAYS before capping
print("Max NUMDAYS before capping:", df_clean["NUMDAYS"].max())
print("Total rows in df_clean:", len(df_clean))
```

```python
# Calculate Q1, Q3, IQR, and upper fence for NUMDAYS
Q1_num = df_clean["NUMDAYS"].quantile(0.25)
Q3_num = df_clean["NUMDAYS"].quantile(0.75)
IQR_num = Q3_num - Q1_num
upper_num = Q3_num + 1.5 * IQR_num

print("Q1:", Q1_num, "| Q3:", Q3_num, "| IQR:", IQR_num)
print("Upper limit for NUMDAYS:", upper_num)
```

```python
# Actually apply the cap to df_clean["NUMDAYS"]
df_clean["NUMDAYS"] = df_clean["NUMDAYS"].clip(upper=upper_num)

print("Max NUMDAYS after capping:", df_clean["NUMDAYS"].max())
```

  <h3 style="color: #7f1d1d; margin: 0 0 6px 0; font-size: 15px;">Explanation — Step 8: IQR Outlier Capping</h3>
  <p style="margin: 0; font-size: 14px; line-height: 1.6;">
    We find <code>Q1</code>, <code>Q3</code>, <code>IQR</code>, and the upper threshold (<code>Q3 + 1.5 * IQR</code>) for <code>NUMDAYS</code>, then cap extreme high values using <code>.clip(upper=upper_num)</code>. This prevents extreme onset day spikes from distorting charts while keeping every row in <code>df_clean</code>.
  </p>
</div>
